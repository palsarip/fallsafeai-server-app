## 2. Pose Extraction

### `02_pose_extraction.ipynb`

```python
# 02_pose_extraction.ipynb
import os
import cv2
import numpy as np
import pandas as pd
import torch
from pathlib import Path
import matplotlib.pyplot as plt
from tqdm import tqdm
import json
from ultralytics import YOLO

# Global settings
VIDEO_LIMIT = 100  # Limit videos/images to process per dataset (for testing)
CONFIDENCE_THRESHOLD = 0.5  # Confidence threshold for pose detection

# Load pre-trained YOLOv8 pose estimation model
print("Loading YOLOv8 pose model...")
model = YOLO('yolov8n-pose.pt')  # Smaller, faster model

# Data directories based on your folder structure
print("Setting up directory paths...")
ROOT_DIR = Path(".").absolute()  # Current directory
NOTEBOOKS_DIR = ROOT_DIR / "notebooks"  # If you're in the notebooks directory

# Convert string paths to Path objects
FALL_DATA_DIR = Path("../data/le2i").absolute()
WORKOUT_VIDEOS_DIR = Path("../data/workout_videos").absolute()
WORKOUT_IMAGES_DIR = Path("../data/workout_images").absolute()
HUMAN_ACTIVITY_DIR = Path("../data/human_activity").absolute()

# Print path status to help debug directory issues
print(f"Checking data directories:")
print(f"FALL_DATA_DIR: {FALL_DATA_DIR} (exists: {FALL_DATA_DIR.exists()})")
print(f"WORKOUT_VIDEOS_DIR: {WORKOUT_VIDEOS_DIR} (exists: {WORKOUT_VIDEOS_DIR.exists()})")
print(f"HUMAN_ACTIVITY_DIR: {HUMAN_ACTIVITY_DIR} (exists: {HUMAN_ACTIVITY_DIR.exists()})")

# Output directory for saving pose data
OUTPUTS_DIR = Path("./outputs").absolute()
POSE_DATA_DIR = OUTPUTS_DIR / "pose_data"
os.makedirs(POSE_DATA_DIR, exist_ok=True)
print(f"Created output directory: {POSE_DATA_DIR}")

def extract_poses_from_video(video_path, output_filename=None, visualize=False):
    """
    Extract poses from video using YOLO and save keypoints to JSON
    
    Args:
        video_path: Path to video file
        output_filename: Where to save the JSON output
        visualize: Whether to visualize sample frames with poses
    
    Returns:
        List of frame pose data
    """
    video_path_str = str(video_path)
    print(f"Attempting to open video: {video_path_str}")
    
    cap = cv2.VideoCapture(video_path_str)
    if not cap.isOpened():
        print(f"Error: Could not open video file {video_path_str}")
        return None
        
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    
    print(f"Video info: {frame_count} frames, {fps} FPS")
    
    # Container for pose data
    video_pose_data = {
        'video_path': video_path_str,
        'total_frames': frame_count,
        'fps': fps,
        'frames': []
    }
    
    # Sample frames to visualize (if requested)
    if visualize:
        vis_indices = np.linspace(0, frame_count-1, 4, dtype=int)
        vis_frames = []
    
    # Process each frame
    frame_idx = 0
    with tqdm(total=frame_count, desc=f"Processing {os.path.basename(video_path_str)}") as pbar:
        while cap.isOpened() and frame_idx < frame_count:
            success, frame = cap.read()
            if not success:
                print(f"Failed to read frame {frame_idx}")
                break
                
            # Run YOLOv8 pose estimation on the frame
            results = model(frame, conf=CONFIDENCE_THRESHOLD)
            
            # Extract pose data for this frame
            frame_data = {
                'frame_idx': frame_idx,
                'timestamp': frame_idx / fps,
                'people': []
            }
            
            # Process detected people
            if results and len(results) > 0 and hasattr(results[0], 'keypoints') and results[0].keypoints is not None:
                keypoints_data = results[0].keypoints.data  # Get keypoints data
                
                for i, person in enumerate(keypoints_data):
                    if person is not None:
                        # Get keypoints (x, y, confidence)
                        keypoints = person.cpu().numpy().tolist()
                        
                        # Add to frame data
                        person_data = {
                            'person_id': i,
                            'keypoints': keypoints
                        }
                        frame_data['people'].append(person_data)
            
            # Add frame data to video data
            video_pose_data['frames'].append(frame_data)
            
            # Visualize if needed
            if visualize and frame_idx in vis_indices:
                # Get annotated frame
                result_plotted = results[0].plot()
                vis_frames.append(result_plotted)
            
            frame_idx += 1
            pbar.update(1)
    
    cap.release()
    
    # Save to JSON if output filename provided
    if output_filename:
        os.makedirs(os.path.dirname(output_filename), exist_ok=True)
        with open(output_filename, 'w') as f:
            json.dump(video_pose_data, f)
    
    # Visualize if requested
    if visualize and vis_frames:
        fig, axes = plt.subplots(1, len(vis_frames), figsize=(20, 5))
        for i, frame in enumerate(vis_frames):
            axes[i].imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
            axes[i].set_title(f"Frame {vis_indices[i]}")
            axes[i].axis('off')
        plt.tight_layout()
        plt.show()
    
    return video_pose_data

def process_video_dataset(video_dir, file_extension, output_subdir, limit=VIDEO_LIMIT):
    """Process videos from a directory and extract poses"""
    # Ensure video_dir is a Path object
    video_dir = Path(video_dir)
    
    # Print current directory details
    print(f"\nProcessing dataset: {output_subdir}")
    print(f"Looking for .{file_extension} files in: {video_dir}")
    print(f"Directory exists: {video_dir.exists()}")
    
    if not video_dir.exists():
        print(f"Warning: Directory {video_dir} does not exist!")
        # Try an alternative approach if the path doesn't exist
        alt_path = ROOT_DIR / "data" / output_subdir
        print(f"Trying alternative path: {alt_path}")
        if alt_path.exists():
            video_dir = alt_path
            print(f"Using alternative path: {video_dir}")
        else:
            print(f"Alternative path doesn't exist either. Skipping dataset.")
            return []
    
    # List directory contents to debug
    print("Directory contents:")
    try:
        for item in video_dir.iterdir():
            print(f"  {item.name} ({'dir' if item.is_dir() else 'file'})")
    except Exception as e:
        print(f"Error listing directory contents: {e}")
    
    # Find videos with the extension
    try:
        videos = list(video_dir.rglob(f"*.{file_extension}"))
        print(f"Found {len(videos)} videos with extension .{file_extension}")
    
        if len(videos) == 0:
            print(f"Trying case-insensitive search...")
            # Try case insensitive match
            videos = []
            for item in video_dir.rglob("*"):
                if item.is_file() and item.suffix.lower() == f".{file_extension.lower()}":
                    videos.append(item)
            print(f"Found {len(videos)} videos with case-insensitive search")
    except Exception as e:
        print(f"Error searching for videos: {e}")
        videos = []
    
    # Limit the number of videos to process (for testing)
    if limit and limit < len(videos):
        videos = videos[:limit]
        print(f"Processing first {limit} videos...")
    
    # Create output directory
    output_dir = POSE_DATA_DIR / output_subdir
    os.makedirs(output_dir, exist_ok=True)
    
    # Process each video
    pose_data_list = []
    for video_path in tqdm(videos, desc=f"Processing {output_subdir} videos"):
        try:
            # Create output path with subdirectories matching the input structure
            try:
                rel_path = video_path.relative_to(video_dir)
            except ValueError:
                # Fallback if relative_to fails
                rel_path = Path(os.path.basename(video_path))
                
            output_path = output_dir / rel_path.with_suffix('.json')
            
            # Create parent directories if they don't exist
            os.makedirs(output_path.parent, exist_ok=True)
            
            # Extract poses with visualization for the first few videos
            visualize = len(pose_data_list) < 3  # Visualize first 3 videos
            pose_data = extract_poses_from_video(
                video_path, 
                output_filename=str(output_path),
                visualize=visualize
            )
            
            if pose_data:
                pose_data_list.append(pose_data)
        except Exception as e:
            print(f"Error processing {video_path}: {str(e)}")
            continue
    
    return pose_data_list

def extract_poses_from_image(image_path, output_filename=None, visualize=False):
    """
    Extract poses from an image using YOLO and save keypoints to JSON
    
    Args:
        image_path: Path to image file
        output_filename: Where to save the JSON output
        visualize: Whether to visualize the image with poses
    
    Returns:
        Image pose data
    """
    image_path_str = str(image_path)
    print(f"Processing image: {image_path_str}")
    
    # Read the image
    img = cv2.imread(image_path_str)
    if img is None:
        print(f"Error: Could not read image file {image_path_str}")
        return None
    
    # Container for pose data
    image_pose_data = {
        'image_path': image_path_str,
        'width': img.shape[1],
        'height': img.shape[0],
        'people': []
    }
    
    # Run YOLOv8 pose estimation on the image
    results = model(img, conf=CONFIDENCE_THRESHOLD)
    
    # Process detected people
    if results and len(results) > 0 and hasattr(results[0], 'keypoints') and results[0].keypoints is not None:
        keypoints_data = results[0].keypoints.data  # Get keypoints data
        
        for i, person in enumerate(keypoints_data):
            if person is not None:
                # Get keypoints (x, y, confidence)
                keypoints = person.cpu().numpy().tolist()
                
                # Add to pose data
                person_data = {
                    'person_id': i,
                    'keypoints': keypoints
                }
                image_pose_data['people'].append(person_data)
    
    # Save to JSON if output filename provided
    if output_filename:
        os.makedirs(os.path.dirname(output_filename), exist_ok=True)
        with open(output_filename, 'w') as f:
            json.dump(image_pose_data, f)
    
    # Visualize if requested
    if visualize:
        # Create a copy for visualization
        vis_img = img.copy()
        
        # Draw keypoints for all people
        for person in image_pose_data['people']:
            keypoints = person['keypoints']
            
            # Draw keypoints
            for kp in keypoints:
                if len(kp) > 2 and kp[2] > 0.5:  # Check confidence
                    x, y = int(kp[0]), int(kp[1])
                    cv2.circle(vis_img, (x, y), 5, (255, 0, 0), -1)
            
            # Draw connections between keypoints
            connections = [
                (0, 1), (0, 2), (1, 3), (2, 4),  # Face & shoulders
                (5, 6), (5, 7), (7, 9), (6, 8), (8, 10),  # Arms
                (5, 11), (6, 12), (11, 13), (12, 14), (13, 15), (14, 16)  # Legs
            ]
            
            for p1, p2 in connections:
                if p1 < len(keypoints) and p2 < len(keypoints):
                    pt1 = keypoints[p1]
                    pt2 = keypoints[p2]
                    if len(pt1) > 2 and len(pt2) > 2 and pt1[2] > 0.5 and pt2[2] > 0.5:
                        x1, y1 = int(pt1[0]), int(pt1[1])
                        x2, y2 = int(pt2[0]), int(pt2[1])
                        cv2.line(vis_img, (x1, y1), (x2, y2), (0, 255, 0), 2)
        
        # Convert to RGB for display
        vis_img = cv2.cvtColor(vis_img, cv2.COLOR_BGR2RGB)
        
        # Display the image with keypoints
        plt.figure(figsize=(10, 10))
        plt.imshow(vis_img)
        plt.title(f"Pose Detection - {os.path.basename(image_path_str)}")
        plt.axis('off')
        plt.tight_layout()
        plt.show()
    
    return image_pose_data

def process_image_dataset(image_dir, file_extensions, output_subdir, limit=VIDEO_LIMIT):
    """Process images from a directory and extract poses
    
    Args:
        image_dir: Directory containing images
        file_extensions: Tuple of image file extensions to look for
        output_subdir: Subdirectory in POSE_DATA_DIR to save results
        limit: Maximum number of images to process
        
    Returns:
        List of pose data dictionaries
    """
    # Ensure image_dir is a Path object
    image_dir = Path(image_dir)
    
    # Print current directory details
    print(f"\nProcessing dataset: {output_subdir}")
    print(f"Looking for images in: {image_dir}")
    print(f"Directory exists: {image_dir.exists()}")
    
    if not image_dir.exists():
        print(f"Warning: Directory {image_dir} does not exist!")
        return []
    
    # List directory contents
    print("Directory contents:")
    try:
        for item in image_dir.iterdir():
            print(f"  {item.name} ({'dir' if item.is_dir() else 'file'})")
    except Exception as e:
        print(f"Error listing directory contents: {e}")
    
    # Find images with the given extensions
    images = []
    for ext in file_extensions:
        try:
            ext_images = list(image_dir.rglob(f"*.{ext}"))
            print(f"Found {len(ext_images)} .{ext} files")
            images.extend(ext_images)
        except Exception as e:
            print(f"Error finding {ext} files: {e}")
    
    print(f"Found {len(images)} total images")
    
    # Limit the number of images to process
    if limit and limit < len(images):
        images = images[:limit]
        print(f"Processing first {limit} images...")
    
    # Create output directory
    output_dir = POSE_DATA_DIR / output_subdir
    os.makedirs(output_dir, exist_ok=True)
    
    # Process each image
    pose_data_list = []
    for image_path in tqdm(images, desc=f"Processing {output_subdir} images"):
        try:
            # Create output path with category subdirectory
            try:
                # Try to maintain directory structure
                rel_path = image_path.relative_to(image_dir)
                category_dir = output_dir / rel_path.parent
            except ValueError:
                # Fallback to using parent directory name as category
                category = image_path.parent.name
                category_dir = output_dir / category
            
            os.makedirs(category_dir, exist_ok=True)
            
            # Create output filename
            output_path = category_dir / f"{image_path.stem}_pose.json"
            
            # Extract poses with visualization for the first few images
            visualize = len(pose_data_list) < 3  # Visualize first 3 images
            pose_data = extract_poses_from_image(
                image_path, 
                output_filename=str(output_path),
                visualize=visualize
            )
            
            if pose_data and pose_data['people']:
                pose_data_list.append(pose_data)
        except Exception as e:
            print(f"Error processing {image_path}: {str(e)}")
            continue
    
    return pose_data_list

def analyze_pose_quality(pose_data_list):
    """Analyze quality of extracted poses across videos"""
    if not pose_data_list:
        return pd.DataFrame()
        
    quality_metrics = []
    
    for video_data in pose_data_list:
        video_path = video_data['video_path']
        total_frames = video_data['total_frames']
        frames_with_people = sum(1 for frame in video_data['frames'] if frame['people'])
        total_detections = sum(len(frame['people']) for frame in video_data['frames'])
        
        # Calculate metrics
        detection_rate = frames_with_people / total_frames if total_frames > 0 else 0
        avg_people_per_frame = total_detections / frames_with_people if frames_with_people > 0 else 0
        
        # Calculate average keypoint confidence across all detections
        confidences = []
        for frame in video_data['frames']:
            for person in frame['people']:
                for kp in person['keypoints']:
                    if len(kp) > 2:  # Check if confidence value exists
                        confidences.append(kp[2])
        
        avg_confidence = np.mean(confidences) if confidences else 0
        
        # Add to metrics
        quality_metrics.append({
            'video': os.path.basename(video_path),
            'total_frames': total_frames,
            'frames_with_detections': frames_with_people,
            'detection_rate': detection_rate,
            'avg_people_per_frame': avg_people_per_frame,
            'avg_keypoint_confidence': avg_confidence
        })
    
    return pd.DataFrame(quality_metrics)

def analyze_image_pose_quality(pose_data_list):
    """Analyze quality of extracted poses across images"""
    if not pose_data_list:
        return pd.DataFrame()
        
    quality_metrics = []
    
    for image_data in pose_data_list:
        image_path = image_data['image_path']
        people_count = len(image_data['people'])
        
        # Calculate average keypoint confidence across all detections
        confidences = []
        for person in image_data['people']:
            for kp in person['keypoints']:
                if len(kp) > 2:  # Check if confidence value exists
                    confidences.append(kp[2])
        
        avg_confidence = np.mean(confidences) if confidences else 0
        
        # Add to metrics
        quality_metrics.append({
            'image': os.path.basename(image_path),
            'width': image_data['width'],
            'height': image_data['height'],
            'people_count': people_count,
            'avg_keypoint_confidence': avg_confidence
        })
    
    return pd.DataFrame(quality_metrics)

def visualize_poses_over_time(pose_data, num_frames=8):
    """Visualize pose changes over time for a single video"""
    if not pose_data:
        print("No pose data provided for visualization")
        return
        
    # Check if this is video or image data
    if 'frames' in pose_data:
        # This is video data
        frames = pose_data['frames']
        total_frames = len(frames)
        
        if total_frames == 0:
            print("No frames in pose data")
            return
        
        # Select evenly spaced frames
        indices = np.linspace(0, total_frames-1, num_frames, dtype=int)
        selected_frames = [frames[i] for i in indices]
        
        # Create a figure for visualization
        fig, axes = plt.subplots(1, num_frames, figsize=(20, 4))
        if num_frames == 1:
            axes = [axes]  # Make it iterable for single frame case
        fig.suptitle(f"Pose Tracking - {os.path.basename(pose_data['video_path'])}")
        
        # Load video to get original frames
        cap = cv2.VideoCapture(pose_data['video_path'])
        if not cap.isOpened():
            print(f"Error: Could not open video file {pose_data['video_path']}")
            return
        
        for i, frame_data in enumerate(selected_frames):
            frame_idx = frame_data['frame_idx']
            cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)
            ret, img = cap.read()
            
            if ret:
                # Convert to RGB for display
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                
                # Draw keypoints for all people in this frame
                for person in frame_data['people']:
                    keypoints = person['keypoints']
                    
                    # Draw keypoints
                    for kp in keypoints:
                        if len(kp) > 2 and kp[2] > 0.5:  # Check confidence
                            x, y = int(kp[0]), int(kp[1])
                            cv2.circle(img, (x, y), 5, (255, 0, 0), -1)
                    
                    # Draw connections between keypoints (simplified)
                    # COCO format connections
                    connections = [
                        (0, 1), (0, 2), (1, 3), (2, 4),  # Face & shoulders
                        (5, 6), (5, 7), (7, 9), (6, 8), (8, 10),  # Arms
                        (5, 11), (6, 12), (11, 13), (12, 14), (13, 15), (14, 16)  # Legs
                    ]
                    
                    for p1, p2 in connections:
                        if p1 < len(keypoints) and p2 < len(keypoints):
                            pt1 = keypoints[p1]
                            pt2 = keypoints[p2]
                            if len(pt1) > 2 and len(pt2) > 2 and pt1[2] > 0.5 and pt2[2] > 0.5:
                                x1, y1 = int(pt1[0]), int(pt1[1])
                                x2, y2 = int(pt2[0]), int(pt2[1])
                                cv2.line(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
                
                # Display the frame with keypoints
                axes[i].imshow(img)
                axes[i].set_title(f"Frame {frame_idx}")
                axes[i].axis('off')
            else:
                axes[i].text(0.5, 0.5, "Frame not available", 
                             horizontalalignment='center', verticalalignment='center')
                axes[i].axis('off')
        
        cap.release()
    else:
        # This is image data - just display the image
        plt.figure(figsize=(10, 10))
        
        # Load the image
        img = cv2.imread(pose_data['image_path'])
        if img is None:
            print(f"Error: Could not read image file {pose_data['image_path']}")
            return
        
        # Convert to RGB for display
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        
        # Draw keypoints for all people
        for person in pose_data['people']:
            keypoints = person['keypoints']
            
            # Draw keypoints
            for kp in keypoints:
                if len(kp) > 2 and kp[2] > 0.5:  # Check confidence
                    x, y = int(kp[0]), int(kp[1])
                    cv2.circle(img, (x, y), 5, (255, 0, 0), -1)
            
            # Draw connections between keypoints
            connections = [
                (0, 1), (0, 2), (1, 3), (2, 4),  # Face & shoulders
                (5, 6), (5, 7), (7, 9), (6, 8), (8, 10),  # Arms
                (5, 11), (6, 12), (11, 13), (12, 14), (13, 15), (14, 16)  # Legs
            ]
            
            for p1, p2 in connections:
                if p1 < len(keypoints) and p2 < len(keypoints):
                    pt1 = keypoints[p1]
                    pt2 = keypoints[p2]
                    if len(pt1) > 2 and len(pt2) > 2 and pt1[2] > 0.5 and pt2[2] > 0.5:
                        x1, y1 = int(pt1[0]), int(pt1[1])
                        x2, y2 = int(pt2[0]), int(pt2[1])
                        cv2.line(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
        
        # Display the image with keypoints
        plt.imshow(img)
        plt.title(f"Pose Detection - {os.path.basename(pose_data['image_path'])}")
        plt.axis('off')
    
    plt.tight_layout()
    plt.show()

# Try to determine the project structure by looking for common directories
def find_data_directory():
    """Look for the data directory in common locations"""
    possible_paths = [
        Path("./data"),                 # Current directory
        Path("../data"),                # One level up
        Path("../../data"),             # Two levels up
        Path("../../../data"),          # Three levels up
        ROOT_DIR / "data",              # Project root
        ROOT_DIR.parent / "data"        # One level up from project root
    ]
    
    for path in possible_paths:
        if path.exists():
            print(f"Found data directory at: {path.absolute()}")
            return path
    
    print("Could not find data directory in common locations")
    return None

# Try to automatically locate data directory
data_dir = find_data_directory()
if data_dir:
    # Update paths if data directory was found
    FALL_DATA_DIR = data_dir / "le2i"
    WORKOUT_VIDEOS_DIR = data_dir / "workout_videos"
    WORKOUT_IMAGES_DIR = data_dir / "workout_images"
    HUMAN_ACTIVITY_DIR = data_dir / "human_activity"
    
    print(f"Updated data paths based on discovered data directory:")
    print(f"FALL_DATA_DIR: {FALL_DATA_DIR} (exists: {FALL_DATA_DIR.exists()})")
    print(f"WORKOUT_VIDEOS_DIR: {WORKOUT_VIDEOS_DIR} (exists: {WORKOUT_VIDEOS_DIR.exists()})")
    print(f"HUMAN_ACTIVITY_DIR: {HUMAN_ACTIVITY_DIR} (exists: {HUMAN_ACTIVITY_DIR.exists()})")

# Process videos from each dataset
print("Starting pose extraction...")

# Process Le2i fall videos
print("\n--- Processing Fall Videos ---")
fall_pose_data = process_video_dataset(
    FALL_DATA_DIR, 
    "avi",  # Le2i videos are typically .avi
    "falls"
)

# Process workout videos
print("\n--- Processing Workout Videos ---")
workout_pose_data = process_video_dataset(
    WORKOUT_VIDEOS_DIR, 
    "mp4",  # Workout videos are usually .mp4
    "workouts"
)

# Process human activity videos (attempt to find videos first)
print("\n--- Processing Human Activity Videos ---")
# First, find all video files (multiple extensions)
video_extensions = ('mp4', 'avi', 'mov', 'flv', 'wmv')
ha_videos = []

print(f"Searching for video files in: {HUMAN_ACTIVITY_DIR}")
if not Path(HUMAN_ACTIVITY_DIR).exists():
    print(f"Warning: Human activity directory {HUMAN_ACTIVITY_DIR} does not exist!")
else:
    for ext in video_extensions:
        try:
            ext_videos = list(Path(HUMAN_ACTIVITY_DIR).rglob(f"*.{ext}"))
            print(f"Found {len(ext_videos)} .{ext} files")
            ha_videos.extend(ext_videos)
        except Exception as e:
            print(f"Error finding {ext} files: {e}")

print(f"Found {len(ha_videos)} total human activity videos")

# Process videos if found
ha_video_pose_data = []
if ha_videos:
    # Limit the number of videos for testing
    ha_videos_to_process = ha_videos[:VIDEO_LIMIT] if len(ha_videos) > VIDEO_LIMIT else ha_videos
    print(f"Processing {len(ha_videos_to_process)} human activity videos...")

    # Process each human activity video
    for video_path in tqdm(ha_videos_to_process, desc="Processing human activity videos"):
        try:
            # Create output path with category subdirectory
            category = video_path.parent.name
            output_dir = POSE_DATA_DIR / "activities" / category
            os.makedirs(output_dir, exist_ok=True)
            
            video_basename = video_path.stem
            output_path = output_dir / f"{video_basename}_poses.json"
            
            # Extract poses with visualization for the first video of each category
            processed_categories = set()
            for v in ha_video_pose_data:
                parts = v['video_path'].replace('\\', '/').split('/')
                if len(parts) >= 2:
                    processed_categories.add(parts[-2])
                    
            visualize = category not in processed_categories
            
            pose_data = extract_poses_from_video(
                video_path, 
                output_filename=str(output_path),
                visualize=visualize
            )
            
            if pose_data:
                ha_video_pose_data.append(pose_data)
        except Exception as e:
            print(f"Error processing {video_path}: {e}")
            continue

    # Process human activity images
print("\n--- Processing Human Activity Images ---")
image_extensions = ('jpg', 'jpeg', 'png')
ha_image_pose_data = process_image_dataset(
    HUMAN_ACTIVITY_DIR,
    image_extensions,
    "activities"
)

# Combine video and image pose data for human activities
ha_pose_data = ha_video_pose_data + ha_image_pose_data
print(f"Extracted pose data from {len(ha_video_pose_data)} videos and {len(ha_image_pose_data)} images")

# Analyze pose quality for each dataset
print("\n--- Analyzing Pose Quality ---")
fall_quality = analyze_pose_quality(fall_pose_data)
workout_quality = analyze_pose_quality(workout_pose_data)

# Use appropriate analysis function for each data type
if ha_video_pose_data and not ha_image_pose_data:
    # Only videos
    activity_quality = analyze_pose_quality(ha_video_pose_data)
elif ha_image_pose_data and not ha_video_pose_data:
    # Only images
    activity_quality = analyze_image_pose_quality(ha_image_pose_data)
else:
    # Combine results from both
    activity_video_quality = analyze_pose_quality(ha_video_pose_data) if ha_video_pose_data else pd.DataFrame()
    activity_image_quality = analyze_image_pose_quality(ha_image_pose_data) if ha_image_pose_data else pd.DataFrame()
    
    # Create combined quality metrics if both exist
    activity_quality = pd.DataFrame()
    if not activity_video_quality.empty:
        activity_quality = activity_video_quality.copy()
        activity_quality['data_type'] = 'video'
    
    if not activity_image_quality.empty:
        # Rename columns to match if needed
        if 'image' in activity_image_quality.columns:
            activity_image_quality = activity_image_quality.rename(columns={'image': 'file'})
        
        image_quality_with_type = activity_image_quality.copy()
        image_quality_with_type['data_type'] = 'image'
        
        if not activity_quality.empty:
            # Combine with existing quality metrics
            activity_quality = pd.concat([activity_quality, image_quality_with_type], ignore_index=True)
        else:
            activity_quality = image_quality_with_type

# Display quality metrics
if not fall_quality.empty:
    print("Fall Detection Pose Quality:")
    print(fall_quality.describe())
    
    # Plot detection rates
    plt.figure(figsize=(10, 5))
    plt.hist(fall_quality['detection_rate'], bins=10)
    plt.title('Fall Videos - Detection Rate')
    plt.xlabel('Detection Rate')
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()

if not workout_quality.empty:
    print("\nWorkout Pose Quality:")
    print(workout_quality.describe())
    
    # Plot detection rates
    plt.figure(figsize=(10, 5))
    plt.hist(workout_quality['detection_rate'], bins=10)
    plt.title('Workout Videos - Detection Rate')
    plt.xlabel('Detection Rate')
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()

if not activity_quality.empty:
    print("\nHuman Activity Pose Quality:")
    print(activity_quality.describe())
    
    # Plot metrics if appropriate columns exist
    if 'detection_rate' in activity_quality.columns:
        plt.figure(figsize=(10, 5))
        plt.hist(activity_quality['detection_rate'], bins=10)
        plt.title('Human Activity - Detection Rate')
        plt.xlabel('Detection Rate')
        plt.ylabel('Count')
        plt.tight_layout()
        plt.show()
    elif 'avg_keypoint_confidence' in activity_quality.columns:
        plt.figure(figsize=(10, 5))
        plt.hist(activity_quality['avg_keypoint_confidence'], bins=10)
        plt.title('Human Activity - Keypoint Confidence')
        plt.xlabel('Average Confidence')
        plt.ylabel('Count')
        plt.tight_layout()
        plt.show()

# Visualize a sample from each category
print("\n--- Visualizing Sample Poses ---")
if fall_pose_data:
    print("Sample Fall Video Poses:")
    visualize_poses_over_time(fall_pose_data[0])

if workout_pose_data:
    print("\nSample Workout Video Poses:")
    # Find a sample with good detection
    good_workout = next((p for p in workout_pose_data if any(len(f['people']) > 0 for f in p['frames'])), None)
    if good_workout:
        visualize_poses_over_time(good_workout)
    else:
        print("No good workout poses found for visualization")

# Visualize human activity samples
if ha_video_pose_data:
    print("\nSample Human Activity Video Poses:")
    # Find a sample with good detection
    good_activity_video = next((p for p in ha_video_pose_data if any(len(f['people']) > 0 for f in p['frames'])), None)
    if good_activity_video:
        visualize_poses_over_time(good_activity_video)
    else:
        print("No good activity video poses found for visualization")

if ha_image_pose_data:
    print("\nSample Human Activity Image Poses:")
    # Find a sample with good detection
    good_activity_image = next((p for p in ha_image_pose_data if p['people']), None)
    if good_activity_image:
        visualize_poses_over_time(good_activity_image)
    else:
        print("No good activity image poses found for visualization")

# Create summary of all processed videos and images
file_paths_by_type = {
    'fall_videos': [data['video_path'] for data in fall_pose_data] if fall_pose_data else [],
    'workout_videos': [data['video_path'] for data in workout_pose_data] if workout_pose_data else [],
    'activity_videos': [data['video_path'] for data in ha_video_pose_data] if ha_video_pose_data else [],
    'activity_images': [data['image_path'] for data in ha_image_pose_data] if ha_image_pose_data else []
}

summary = {
    'fall_videos': {
        'count': len(fall_pose_data),
        'file_paths': file_paths_by_type['fall_videos'],
        'quality_metrics': fall_quality.to_dict('records') if not fall_quality.empty else []
    },
    'workout_videos': {
        'count': len(workout_pose_data),
        'file_paths': file_paths_by_type['workout_videos'],
        'quality_metrics': workout_quality.to_dict('records') if not workout_quality.empty else []
    },
    'activity_data': {
        'total_count': len(ha_pose_data),
        'video_count': len(ha_video_pose_data),
        'image_count': len(ha_image_pose_data),
        'video_paths': file_paths_by_type['activity_videos'],
        'image_paths': file_paths_by_type['activity_images'],
        'quality_metrics': activity_quality.to_dict('records') if not activity_quality.empty else []
    }
}

# Save summary to JSON
summary_path = POSE_DATA_DIR / 'pose_extraction_summary.json'
with open(summary_path, 'w') as f:
    json.dump(summary, f, indent=2)

print(f"\nPose extraction complete! Data saved to {POSE_DATA_DIR}")
print(f"Summary saved to {summary_path}")
```
