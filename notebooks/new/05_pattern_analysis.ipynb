```python
# 05_pattern_analysis_tensorflow.ipynb
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import json
import pickle
import tensorflow as tf
from sklearn.metrics import confusion_matrix, classification_report, roc_curve, auc, precision_recall_curve
from sklearn.preprocessing import StandardScaler
from sklearn.inspection import permutation_importance
import shap

# Set up paths
ROOT_DIR = Path(".")
FEATURES_DIR = ROOT_DIR / "outputs" / "features"
MODELS_DIR = ROOT_DIR / "outputs" / "models"
ANALYSIS_DIR = ROOT_DIR / "outputs" / "analysis"
os.makedirs(ANALYSIS_DIR, exist_ok=True)

class FallDetectionEnsemble:
    def __init__(self, tf_model, trad_model, scaler):
        self.tf_model = tf_model
        self.trad_model = trad_model
        self.scaler = scaler
    
    def predict(self, X):
        # Scale data for TensorFlow model
        X_scaled = self.scaler.transform(X)
        
        # Get predictions from both models
        tf_pred_prob = self.tf_model.predict(X_scaled)
        trad_pred_prob = self.trad_model.predict_proba(X)[:, 1]
        
        # Average the probabilities (weighted)
        ensemble_pred_prob = 0.7 * tf_pred_prob.flatten() + 0.3 * trad_pred_prob
        
        # Convert to binary predictions
        return (ensemble_pred_prob >= 0.5).astype(int)
    
    def predict_proba(self, X):
        # Scale data for TensorFlow model
        X_scaled = self.scaler.transform(X)
        
        # Get predictions from both models
        tf_pred_prob = self.tf_model.predict(X_scaled).flatten()
        trad_pred_prob = self.trad_model.predict_proba(X)[:, 1]
        
        # Average the probabilities (weighted)
        ensemble_pred_prob = 0.7 * tf_pred_prob + 0.3 * trad_pred_prob
        
        # Format for sklearn compatibility
        return np.column_stack((1 - ensemble_pred_prob, ensemble_pred_prob))

# TensorFlow model wrapper for scikit-learn compatibility
class TensorFlowWrapper:
    def __init__(self, model, scaler):
        self.model = model
        self.scaler = scaler
        
    def fit(self, X, y):
        # This is just a placeholder to comply with the scikit-learn API
        return self
        
    def predict(self, X):
        # Scale the input data
        X_scaled = self.scaler.transform(X)
        # Get predictions and convert to binary
        preds = self.model.predict(X_scaled)
        return (preds > 0.5).astype(int).flatten()
    
    def score(self, X, y):
        # Get predictions
        y_pred = self.predict(X)
        # Calculate accuracy score
        from sklearn.metrics import accuracy_score
        return accuracy_score(y, y_pred)

print("Fall Detection - TensorFlow Pattern Analysis")

# 1. Load Model and Data
print("\n1. Loading Model and Data...")

# Check if TensorFlow model exists
tf_model_path = MODELS_DIR / "tensorflow" / "best_model.keras"
trad_model_path = MODELS_DIR / "fall_detection_traditional_model.pkl"
ensemble_model_path = MODELS_DIR / "ensemble_model.pkl"

# Load model info to determine which model to use
model_info_path = MODELS_DIR / "model_info.json"
if model_info_path.exists():
    with open(model_info_path, 'r') as f:
        model_info = json.load(f)
    print(f"Using model type: {model_info.get('model_type', 'unknown')}")
    model_type = model_info.get('model_type', 'tensorflow')
else:
    print("Model info not found, defaulting to TensorFlow model")
    model_type = 'tensorflow'

# Load appropriate model
if model_type == 'tensorflow' and tf_model_path.exists():
    print(f"Loading TensorFlow model from {tf_model_path}")
    model = tf.keras.models.load_model(tf_model_path)
    model_name = "TensorFlow"
elif model_type == 'ensemble' and ensemble_model_path.exists():
    print(f"Loading ensemble model from {ensemble_model_path}")
    with open(ensemble_model_path, 'rb') as f:
        model = pickle.load(f)
    model_name = "Ensemble"
else:
    print(f"Loading traditional model from {trad_model_path}")
    with open(trad_model_path, 'rb') as f:
        model = pickle.load(f)
    model_name = "Traditional ML"

# Load feature data
def load_feature_data():
    """Load feature data from CSV"""
    # Try to load normalized features first
    normalized_file = FEATURES_DIR / "normalized_features.csv"
    if normalized_file.exists():
        print(f"Loading normalized features from: {normalized_file}")
        df = pd.read_csv(normalized_file)
    else:
        # Fall back to raw features
        raw_file = FEATURES_DIR / "features_for_modeling.csv"
        if raw_file.exists():
            print(f"Loading raw features from: {raw_file}")
            df = pd.read_csv(raw_file)
        else:
            raise FileNotFoundError("No feature data found!")
    
    print(f"Loaded feature data with {df.shape[0]} samples and {df.shape[1]} features")
    return df

# Load feature data
feature_df = load_feature_data()

# Separate features and labels
X = feature_df.drop(['label', 'file_path', 'data_type'], axis=1, errors='ignore')
if 'label' in feature_df.columns:
    if feature_df['label'].dtype == object:
        y = feature_df['label'].map({'fall': 1, 'workout': 0})
    else:
        y = feature_df['label']
else:
    raise ValueError("Label column not found in feature data")

# Split into train/test if needed for analysis
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Scale features for TensorFlow model
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 2. Identify Distinctive Patterns Between Falls and Workouts
print("\n2. Identifying Distinctive Patterns Between Falls and Workouts...")

# Function to plot feature distributions by label
def plot_feature_distributions(df, feature_names, figsize=(15, 10)):
    """Plot distribution of features by class label"""
    num_features = len(feature_names)
    num_rows = (num_features + 2) // 3  # Calculate rows needed (3 columns)
    
    fig, axes = plt.subplots(num_rows, 3, figsize=figsize)
    axes = axes.flatten()
    
    for i, feature in enumerate(feature_names):
        if i < len(axes):
            sns.boxplot(x='label', y=feature, data=df, ax=axes[i])
            axes[i].set_title(f'Distribution of {feature}')
            axes[i].set_xlabel('')
            if df['label'].dtype == object:
                axes[i].set_xticklabels(['Non-Fall', 'Fall'])
            else:
                axes[i].set_xticklabels(['Non-Fall', 'Fall'])
    
    # Hide unused subplots
    for i in range(num_features, len(axes)):
        axes[i].set_visible(False)
    
    plt.tight_layout()
    plt.suptitle('Feature Distributions: Falls vs Non-Falls', y=1.02)
    plt.savefig(ANALYSIS_DIR / 'feature_distributions.png')
    plt.show()
    return fig

# Create a feature importance analysis based on model type
if model_type == 'tensorflow':
    # For TensorFlow models, create a wrapper for permutation importance
    class TensorFlowWrapper:
        def __init__(self, model, scaler):
            self.model = model
            self.scaler = scaler
            
        def fit(self, X, y):
            return self
            
        def predict(self, X):
            X_scaled = self.scaler.transform(X)
            preds = self.model.predict(X_scaled)
            return (preds > 0.5).astype(int).flatten()
        
        def score(self, X, y):
            from sklearn.metrics import accuracy_score
            y_pred = self.predict(X)
            return accuracy_score(y, y_pred)
    
    # Create wrapper
    tf_wrapper = TensorFlowWrapper(model, scaler)
    
    # Calculate permutation importance
    perm_importance = permutation_importance(
        tf_wrapper, X_test, y_test,
        n_repeats=5, random_state=42
    )
    
    # Create importance DataFrame
    importance_df = pd.DataFrame({
        'Feature': X.columns,
        'Importance': perm_importance.importances_mean,
        'Std': perm_importance.importances_std
    }).sort_values('Importance', ascending=False)
    
    # Plot importance
    plt.figure(figsize=(12, 8))
    top_features = importance_df.head(15)
    sns.barplot(x='Importance', y='Feature', data=top_features)
    plt.title(f'Top 15 Features - TensorFlow Model')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'feature_importance_tensorflow.png')
    plt.show()
    
    print("Top 10 most important features:")
    print(importance_df[['Feature', 'Importance']].head(10).to_string(index=False))
    
    # Get the most important features for further analysis
    top_features_list = importance_df.head(10)['Feature'].tolist()
    
elif model_type == 'ensemble':
    # For ensemble model, use manual prediction to identify feature importance
    # Since we can't directly extract feature importance from the ensemble,
    # we'll analyze the constituent models
    
    # Create a feature importance plot by comparing prediction changes
    # when features are permuted
    if hasattr(model, 'tf_model') and hasattr(model, 'trad_model'):
        # If we can access the constituent models, analyze them separately
        print("Analyzing individual models in the ensemble...")
        
        # For TensorFlow model
        tf_wrapper = TensorFlowWrapper(model.tf_model, scaler)
        perm_importance_tf = permutation_importance(
            tf_wrapper, X_test, y_test,
            n_repeats=5, random_state=42
        )
        
        # Create importance DataFrame for TensorFlow
        importance_df_tf = pd.DataFrame({
            'Feature': X.columns,
            'Importance': perm_importance_tf.importances_mean,
            'Std': perm_importance_tf.importances_std
        }).sort_values('Importance', ascending=False)
        
        # For traditional model
        perm_importance_trad = permutation_importance(
            model.trad_model, X_test, y_test,
            n_repeats=5, random_state=42
        )
        
        # Create importance DataFrame for traditional
        importance_df_trad = pd.DataFrame({
            'Feature': X.columns,
            'Importance': perm_importance_trad.importances_mean,
            'Std': perm_importance_trad.importances_std
        }).sort_values('Importance', ascending=False)
        
        # Combine results to get ensemble importance
        importance_df = pd.DataFrame({
            'Feature': X.columns,
            'TF_Importance': importance_df_tf['Importance'],
            'Trad_Importance': importance_df_trad['Importance'],
            'Ensemble_Importance': 0.7 * importance_df_tf['Importance'] + 0.3 * importance_df_trad['Importance']
        }).sort_values('Ensemble_Importance', ascending=False)
        
        # Plot importance
        plt.figure(figsize=(12, 8))
        top_features = importance_df.head(15)
        sns.barplot(x='Ensemble_Importance', y='Feature', data=top_features)
        plt.title(f'Top 15 Features - Ensemble Model')
        plt.tight_layout()
        plt.savefig(ANALYSIS_DIR / 'feature_importance_ensemble.png')
        plt.show()
        
        print("Top 10 most important features (Ensemble):")
        print(importance_df[['Feature', 'Ensemble_Importance']].head(10).to_string(index=False))
        
        # Get the most important features for further analysis
        top_features_list = importance_df.head(10)['Feature'].tolist()
    else:
        # If we can't access the constituent models, analyze as a black box
        perm_importance = permutation_importance(
            model, X_test, y_test,
            n_repeats=5, random_state=42
        )
        
        # Create importance DataFrame
        importance_df = pd.DataFrame({
            'Feature': X.columns,
            'Importance': perm_importance.importances_mean,
            'Std': perm_importance.importances_std
        }).sort_values('Importance', ascending=False)
        
        # Plot importance
        plt.figure(figsize=(12, 8))
        top_features = importance_df.head(15)
        sns.barplot(x='Importance', y='Feature', data=top_features)
        plt.title(f'Top 15 Features - Ensemble Model')
        plt.tight_layout()
        plt.savefig(ANALYSIS_DIR / 'feature_importance_ensemble.png')
        plt.show()
        
        print("Top 10 most important features:")
        print(importance_df[['Feature', 'Importance']].head(10).to_string(index=False))
        
        # Get the most important features for further analysis
        top_features_list = importance_df.head(10)['Feature'].tolist()
else:
    # For traditional ML models
    if hasattr(model, 'named_steps') and hasattr(model.named_steps.get('classifier', None), 'feature_importances_'):
        # For tree-based models
        feature_importances = model.named_steps['classifier'].feature_importances_
        importance_df = pd.DataFrame({
            'Feature': X.columns,
            'Importance': feature_importances
        }).sort_values('Importance', ascending=False)
    elif hasattr(model, 'named_steps') and hasattr(model.named_steps.get('classifier', None), 'coef_'):
        # For linear models
        coef = model.named_steps['classifier'].coef_[0]
        importance_df = pd.DataFrame({
            'Feature': X.columns,
            'Coefficient': coef,
            'Importance': np.abs(coef)
        }).sort_values('Importance', ascending=False)
    else:
        # Fallback to permutation importance
        perm_importance = permutation_importance(
            model, X_test, y_test,
            n_repeats=5, random_state=42
        )
        
        importance_df = pd.DataFrame({
            'Feature': X.columns,
            'Importance': perm_importance.importances_mean,
            'Std': perm_importance.importances_std
        }).sort_values('Importance', ascending=False)
    
    # Plot importance
    plt.figure(figsize=(12, 8))
    top_features = importance_df.head(15)
    
    if 'Coefficient' in importance_df.columns:
        sns.barplot(x='Coefficient', y='Feature', data=top_features)
        plt.title(f'Top 15 Features - {model_name} (Coefficients)')
    else:
        sns.barplot(x='Importance', y='Feature', data=top_features)
        plt.title(f'Top 15 Features - {model_name}')
    
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'feature_importance_traditional.png')
    plt.show()
    
    print("Top 10 most important features:")
    if 'Coefficient' in importance_df.columns:
        print(importance_df[['Feature', 'Coefficient']].head(10).to_string(index=False))
    else:
        print(importance_df[['Feature', 'Importance']].head(10).to_string(index=False))
    
    # Get the most important features for further analysis
    top_features_list = importance_df.head(10)['Feature'].tolist()

# Select top features for distribution analysis
top_features_list = top_features_list[:8]  # Limit to 8 for visualization clarity
plot_feature_distributions(feature_df, top_features_list)

# Create a DataFrame with features and label for analysis
analysis_df = pd.DataFrame(X[top_features_list])
analysis_df['label'] = y

# Calculate correlation between features
correlation_matrix = analysis_df.corr()

# Plot correlation heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.savefig(ANALYSIS_DIR / 'feature_correlation.png')
plt.show()

# Create a pairplot for top 5 features
if len(top_features_list) >= 5:
    top_5_features = top_features_list[:5]
    pair_df = analysis_df[top_5_features + ['label']].copy()
    if pair_df['label'].dtype != object:
        pair_df['label'] = pair_df['label'].map({0: 'Non-Fall', 1: 'Fall'})
    
    print("Creating pairplot for top 5 features...")
    pair_plot = sns.pairplot(pair_df, hue='label', diag_kind='kde', 
                           plot_kws={'alpha': 0.6})
    pair_plot.fig.suptitle('Pairwise Relationships Between Top Features', y=1.02)
    pair_plot.savefig(ANALYSIS_DIR / 'feature_pairplot.png')
    plt.show()

# 3. Analyze False Positives and False Negatives
print("\n3. Analyzing False Positives and False Negatives...")

# Get model predictions
if model_type == 'tensorflow':
    y_pred_prob = model.predict(X_test_scaled)
    y_pred = (y_pred_prob > 0.5).astype(int).flatten()
    if y_pred.ndim > 1 and y_pred.shape[1] == 1:
        y_pred = y_pred.flatten()
elif model_type == 'ensemble':
    # For ensemble models
    X_test_scaled = scaler.transform(X_test)
    tf_pred_prob = model.tf_model.predict(X_test_scaled).flatten()
    trad_pred_prob = model.trad_model.predict_proba(X_test)[:, 1]
    y_pred_prob = 0.7 * tf_pred_prob + 0.3 * trad_pred_prob
    y_pred = (y_pred_prob >= 0.5).astype(int)
else:
    # For traditional ML models
    y_pred = model.predict(X_test)
    y_pred_prob = model.predict_proba(X_test)[:, 1]

# Create a DataFrame for error analysis
error_df = pd.DataFrame({
    'true_label': y_test,
    'predicted_label': y_pred,
    'probability': y_pred_prob if y_pred_prob is not None else np.nan
})

# Add features for analysis
for feature in top_features_list:
    error_df[feature] = X_test[feature].values

# Identify false positives and false negatives
false_positives = error_df[(error_df['true_label'] == 0) & (error_df['predicted_label'] == 1)]
false_negatives = error_df[(error_df['true_label'] == 1) & (error_df['predicted_label'] == 0)]
true_positives = error_df[(error_df['true_label'] == 1) & (error_df['predicted_label'] == 1)]
true_negatives = error_df[(error_df['true_label'] == 0) & (error_df['predicted_label'] == 0)]

print(f"Correctly classified falls (TP): {len(true_positives)}")
print(f"Correctly classified workouts (TN): {len(true_negatives)}")
print(f"Workouts misclassified as falls (FP): {len(false_positives)}")
print(f"Missed falls (FN): {len(false_negatives)}")

# Calculate confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Plot confusion matrix with percentages
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
           xticklabels=['Non-Fall', 'Fall'],
           yticklabels=['Non-Fall', 'Fall'])
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title(f'Confusion Matrix - {model_name} Model')
plt.savefig(ANALYSIS_DIR / 'confusion_matrix.png')
plt.show()

# Analyze false positives
if len(false_positives) > 0:
    print("\nAnalyzing False Positives (Workouts classified as falls):")
    print(false_positives[top_features_list].describe())
    
    # Compare false positives to true positives
    fp_vs_tp = pd.DataFrame({
        'False Positive': false_positives[top_features_list].mean(),
        'True Positive': true_positives[top_features_list].mean() if len(true_positives) > 0 else pd.Series([np.nan] * len(top_features_list), index=top_features_list)
    })
    
    # Plot comparison
    plt.figure(figsize=(12, 8))
    fp_vs_tp.plot(kind='bar')
    plt.title('Feature Comparison: False Positives vs True Positives (Mean Values)')
    plt.ylabel('Normalized Feature Value')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'false_positives_analysis.png')
    plt.show()

# Analyze false negatives
if len(false_negatives) > 0:
    print("\nAnalyzing False Negatives (Falls classified as workouts):")
    print(false_negatives[top_features_list].describe())
    
    # Compare false negatives to true negatives
    fn_vs_tn = pd.DataFrame({
        'False Negative': false_negatives[top_features_list].mean(),
        'True Negative': true_negatives[top_features_list].mean() if len(true_negatives) > 0 else pd.Series([np.nan] * len(top_features_list), index=top_features_list)
    })
    
    # Plot comparison
    plt.figure(figsize=(12, 8))
    fn_vs_tn.plot(kind='bar')
    plt.title('Feature Comparison: False Negatives vs True Negatives (Mean Values)')
    plt.ylabel('Normalized Feature Value')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'false_negatives_analysis.png')
    plt.show()

# 4. Threshold Tuning for Real-World Application
print("\n4. Threshold Tuning for Real-World Application...")

# Calculate precision and recall at different thresholds
precision, recall, thresholds = precision_recall_curve(y_test, y_pred_prob)

# Calculate F1 scores at different thresholds
f1_scores = 2 * (precision[:-1] * recall[:-1]) / (precision[:-1] + recall[:-1] + 1e-10)

# Calculate F2 scores (weighing recall higher)
beta = 2
f2_scores = (1 + beta**2) * (precision[:-1] * recall[:-1]) / ((beta**2 * precision[:-1]) + recall[:-1] + 1e-10)

# Create DataFrame for threshold analysis
threshold_df = pd.DataFrame({
    'Threshold': thresholds,
    'Precision': precision[:-1],
    'Recall': recall[:-1],
    'F1': f1_scores,
    'F2': f2_scores
})

# Plot precision-recall curve
plt.figure(figsize=(10, 6))
plt.plot(recall, precision, marker='.', label='Precision-Recall curve')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve')
plt.grid(True)
plt.savefig(ANALYSIS_DIR / 'precision_recall_curve.png')
plt.show()

# Plot F1 and F2 scores vs threshold
plt.figure(figsize=(10, 6))
plt.plot(thresholds, f1_scores, label='F1 Score')
plt.plot(thresholds, f2_scores, label='F2 Score (weighs recall higher)')
plt.xlabel('Threshold')
plt.ylabel('Score')
plt.title('F1 and F2 Scores vs. Threshold')
plt.legend()
plt.grid(True)
plt.savefig(ANALYSIS_DIR / 'f_scores_vs_threshold.png')
plt.show()

# Find optimal thresholds
optimal_f1_idx = np.argmax(f1_scores)
optimal_f1_threshold = thresholds[optimal_f1_idx]

optimal_f2_idx = np.argmax(f2_scores)
optimal_f2_threshold = thresholds[optimal_f2_idx]

print(f"Default classification threshold: 0.5")
print(f"Optimal threshold for balanced precision/recall (F1): {optimal_f1_threshold:.4f}")
print(f"Optimal threshold for higher recall (F2): {optimal_f2_threshold:.4f}")

# Calculate confusion matrices for different thresholds
cms = {}
threshold_labels = {
    0.5: 'Default (0.5)',
    optimal_f1_threshold: 'F1 Optimal',
    optimal_f2_threshold: 'F2 Optimal (Recall Focus)'
}

plt.figure(figsize=(15, 5))
for i, (threshold, label) in enumerate(threshold_labels.items()):
    y_pred_custom = (y_pred_prob >= threshold).astype(int)
    cm = confusion_matrix(y_test, y_pred_custom)
    cms[label] = cm
    
    # Plot confusion matrix
    plt.subplot(1, 3, i+1)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
               xticklabels=['Non-Fall', 'Fall'],
               yticklabels=['Non-Fall', 'Fall'])
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.title(f'Threshold: {threshold:.3f} ({label})')

plt.tight_layout()
plt.savefig(ANALYSIS_DIR / 'threshold_confusion_matrices.png')
plt.show()

# Calculate metrics for different thresholds
metrics = []
for threshold, label in threshold_labels.items():
    y_pred_custom = (y_pred_prob >= threshold).astype(int)
    cm = confusion_matrix(y_test, y_pred_custom)
    
    # Calculate metrics
    tn, fp, fn, tp = cm.ravel()
    precision_val = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall_val = tp / (tp + fn) if (tp + fn) > 0 else 0
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    f1 = 2 * (precision_val * recall_val) / (precision_val + recall_val) if (precision_val + recall_val) > 0 else 0
    accuracy = (tp + tn) / (tp + fp + tn + fn)
    
    # For real-world fall detection, add more metrics
    false_alarm_rate = fp / (fp + tn) if (fp + tn) > 0 else 0
    miss_rate = fn / (fn + tp) if (fn + tp) > 0 else 0
    
    metrics.append({
        'Threshold': threshold,
        'Label': label,
        'Accuracy': accuracy,
        'Precision': precision_val,
        'Recall (Sensitivity)': recall_val,
        'Specificity': specificity,
        'F1 Score': f1,
        'False Alarm Rate': false_alarm_rate,
        'Miss Rate': miss_rate,
        'TP': tp,
        'FP': fp,
        'TN': tn,
        'FN': fn
    })

# Create metrics DataFrame
metrics_df = pd.DataFrame(metrics)
print("\nPerformance metrics at different thresholds:")
print(metrics_df[['Label', 'Threshold', 'Accuracy', 'Precision', 'Recall (Sensitivity)', 'False Alarm Rate', 'Miss Rate']])

# Save metrics to CSV
metrics_df.to_csv(ANALYSIS_DIR / 'threshold_metrics.csv', index=False)

# Create a more detailed analysis of the optimal thresholds
plt.figure(figsize=(12, 6))

# Plot precision, recall vs threshold
plt.plot(thresholds, precision[:-1], label='Precision', marker='', linestyle='-')
plt.plot(thresholds, recall[:-1], label='Recall', marker='', linestyle='-')

# Highlight optimal thresholds
plt.axvline(x=0.5, color='gray', linestyle='--', label='Default (0.5)')
plt.axvline(x=optimal_f1_threshold, color='green', linestyle='--', label=f'F1 Optimal ({optimal_f1_threshold:.3f})')
plt.axvline(x=optimal_f2_threshold, color='red', linestyle='--', label=f'F2 Optimal ({optimal_f2_threshold:.3f})')

plt.xlabel('Threshold')
plt.ylabel('Score')
plt.title('Precision & Recall vs. Threshold')
plt.legend()
plt.grid(True)
plt.savefig(ANALYSIS_DIR / 'threshold_analysis.png')
plt.show()

# ROC curve and AUC
fpr, tpr, _ = roc_curve(y_test, y_pred_prob)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(10, 6))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC curve (AUC = {roc_auc:.3f})')
plt.plot([0, 1], [0, 1], color='gray', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend(loc="lower right")
plt.grid(True)
plt.savefig(ANALYSIS_DIR / 'roc_curve.png')
plt.show()

# 5. Fall vs. Workout Motion Pattern Analysis
print("\n5. Fall vs. Workout Motion Pattern Analysis...")

# Select motion-specific features
motion_features = [col for col in X.columns if any(term in col for term in 
                                                  ['velocity', 'accel', 'height', 'collapse',
                                                   'movement', 'aspect_ratio', 'bounding_box'])]

if motion_features:
    # Filter to meaningful features
    motion_features = motion_features[:8]  # Limit to 8 for clarity
    
    # Create a detailed visualization of motion patterns
    plt.figure(figsize=(15, 10))
    for i, feature in enumerate(motion_features):
        if i < 8:  # Ensure we only plot up to 8 features
            plt.subplot(2, 4, i+1)
            if feature in X.columns:
                sns.boxplot(x='label', y=feature, data=pd.DataFrame({
                    'label': y.map({0: 'Non-Fall', 1: 'Fall'}),
                    feature: X[feature]
                }))
                plt.title(f'{feature}')
                plt.xlabel('')
    
    plt.tight_layout()
    plt.suptitle('Motion Features: Falls vs Non-Falls', y=1.02)
    plt.savefig(ANALYSIS_DIR / 'motion_analysis.png')
    plt.show()
    
    # Create a bar chart to visualize fall vs. workout motion patterns
    fall_means = pd.DataFrame(X[y==1][motion_features].mean()).reset_index()
    fall_means.columns = ['Feature', 'Fall']
    nonfall_means = pd.DataFrame(X[y==0][motion_features].mean()).reset_index()
    nonfall_means.columns = ['Feature', 'Non-Fall']
    
    # Merge for comparison
    motion_comparison = pd.merge(fall_means, nonfall_means, on='Feature')
    
    # Calculate ratio
    motion_comparison['Ratio'] = motion_comparison['Fall'] / motion_comparison['Non-Fall']
    
    # Create a bar chart
    plt.figure(figsize=(12, 8))
    motion_long = pd.melt(motion_comparison, id_vars=['Feature'], value_vars=['Fall', 'Non-Fall'])
    sns.barplot(x='Feature', y='value', hue='variable', data=motion_long)
    plt.title('Fall vs. Non-Fall Motion Feature Comparison')
    plt.ylabel('Feature Value')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'fall_pattern_comparison.png')
    plt.show()
    
    # Display the feature ratio
    plt.figure(figsize=(10, 6))
    motion_comparison = motion_comparison.sort_values('Ratio', ascending=False)
    plt.bar(motion_comparison['Feature'], motion_comparison['Ratio'])
    plt.axhline(y=1, color='r', linestyle='--')
    plt.title('Ratio of Fall to Non-Fall Feature Values')
    plt.ylabel('Fall/Non-Fall Ratio')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'feature_ratio.png')
    plt.show()
    
    # Identify the most distinctive fall features
    print("\nMost distinctive fall features (highest fall/non-fall ratio):")
    for i, (_, row) in enumerate(motion_comparison.iterrows()):
        if i < 5:  # Show top 5
            if row['Ratio'] > 1:
                print(f"{row['Feature']}: {row['Ratio']:.2f}x higher in falls")
            else:
                print(f"{row['Feature']}: {1/row['Ratio']:.2f}x lower in falls")

# 6. Feature Interaction Analysis
print("\n6. Feature Interaction Analysis...")

# Select the top features for interaction analysis
if len(top_features_list) >= 4:
    interaction_features = top_features_list[:4]
    
    # Create interaction plots
    fig, axes = plt.subplots(nrows=len(interaction_features), ncols=len(interaction_features), 
                            figsize=(15, 15))
    
    # Add labels for easy identification
    labels = y.map({0: 'Non-Fall', 1: 'Fall'})
    
    # Create scatter plots for each feature pair
    for i, feat1 in enumerate(interaction_features):
        for j, feat2 in enumerate(interaction_features):
            ax = axes[i, j]
            if i == j:  # Diagonal - show distribution
                for label, color in zip([0, 1], ['blue', 'red']):
                    sns.kdeplot(X.loc[y == label, feat1], ax=ax, color=color, label=f"{'Fall' if label==1 else 'Non-Fall'}")
                if i == 0:  # Only add legend to the first plot
                    ax.legend()
            else:  # Off-diagonal - show scatter
                ax.scatter(X[feat2], X[feat1], c=y, cmap='coolwarm', alpha=0.6)
                ax.set_xlabel(feat2)
                ax.set_ylabel(feat1)
    
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'feature_interaction.png')
    plt.show()

# 7. Dimensionality Reduction for Visualization
print("\n7. Dimensionality Reduction for Feature Visualization...")

# Use PCA and t-SNE for dimensionality reduction
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Apply t-SNE
tsne = TSNE(n_components=2, random_state=42)
X_tsne = tsne.fit_transform(X)

# Create visualization DataFrames
pca_df = pd.DataFrame(data=X_pca, columns=['PC1', 'PC2'])
pca_df['label'] = y.map({0: 'Non-Fall', 1: 'Fall'})

tsne_df = pd.DataFrame(data=X_tsne, columns=['TSNE1', 'TSNE2'])
tsne_df['label'] = y.map({0: 'Non-Fall', 1: 'Fall'})

# Plot PCA results
plt.figure(figsize=(10, 8))
sns.scatterplot(x='PC1', y='PC2', hue='label', data=pca_df, palette='Set1', alpha=0.8)
plt.title('PCA: Falls vs. Non-Falls')
plt.xlabel(f'Principal Component 1 (Explained Variance: {pca.explained_variance_ratio_[0]:.2f})')
plt.ylabel(f'Principal Component 2 (Explained Variance: {pca.explained_variance_ratio_[1]:.2f})')
plt.tight_layout()
plt.savefig(ANALYSIS_DIR / 'pca_visualization.png')
plt.show()

# Plot t-SNE results
plt.figure(figsize=(10, 8))
sns.scatterplot(x='TSNE1', y='TSNE2', hue='label', data=tsne_df, palette='Set1', alpha=0.8)
plt.title('t-SNE: Falls vs. Non-Falls')
plt.tight_layout()
plt.savefig(ANALYSIS_DIR / 't-sne_visualization.png')
plt.show()

# 8. Model Interpretation with SHAP (if available)
print("\n8. Model Interpretation with SHAP...")

try:
    if model_type == 'tensorflow':
        # For TensorFlow models, use a wrapper with SHAP
        import shap
        
        # Create a wrapper for SHAP
        def model_predict(X):
            X_scaled = scaler.transform(X)
            return model.predict(X_scaled).flatten()
        
        # Use a subset of data for SHAP analysis (for speed)
        X_sample = X_train.sample(min(100, len(X_train)), random_state=42)
        
        # Initialize the SHAP explainer
        explainer = shap.KernelExplainer(model_predict, X_sample)
        
        # Calculate SHAP values for test data
        X_test_sample = X_test.sample(min(100, len(X_test)), random_state=42)
        shap_values = explainer.shap_values(X_test_sample)
        
        # Summary plot
        plt.figure(figsize=(10, 8))
        shap.summary_plot(shap_values, X_test_sample, plot_type="bar", show=False)
        plt.title('Feature Importance based on SHAP Values')
        plt.tight_layout()
        plt.savefig(ANALYSIS_DIR / 'shap_importance.png')
        plt.show()
        
        # Detailed SHAP plot
        plt.figure(figsize=(12, 10))
        shap.summary_plot(shap_values, X_test_sample, show=False)
        plt.title('SHAP Value Summary')
        plt.tight_layout()
        plt.savefig(ANALYSIS_DIR / 'shap_summary.png')
        plt.show()
        
        # Show SHAP dependence plots for top features
        for feature in top_features_list[:3]:
            plt.figure(figsize=(10, 6))
            shap.dependence_plot(feature, shap_values, X_test_sample, show=False)
            plt.title(f'SHAP Dependence Plot for {feature}')
            plt.tight_layout()
            plt.savefig(ANALYSIS_DIR / f'shap_dependence_{feature}.png')
            plt.show()
    
    else:
        print("SHAP analysis currently optimized for TensorFlow models")
        print("Skipping detailed SHAP interpretation")
except Exception as e:
    print(f"SHAP analysis error: {e}")
    print("Skipping SHAP interpretation")

# 9. Save Findings
print("\n9. Saving Pattern Analysis Findings...")

# Summarize key findings
findings = {
    "model_type": model_type,
    "model_name": model_name,
    "top_features": top_features_list,
    "motion_features": motion_features,
    "optimal_thresholds": {
        "default": 0.5,
        "f1_optimal": float(optimal_f1_threshold),
        "f2_optimal": float(optimal_f2_threshold)
    },
    "performance_metrics": {
        "accuracy": float(metrics_df[metrics_df['Label'] == 'Default (0.5)']['Accuracy'].values[0]),
        "precision": float(metrics_df[metrics_df['Label'] == 'Default (0.5)']['Precision'].values[0]),
        "recall": float(metrics_df[metrics_df['Label'] == 'Default (0.5)']['Recall (Sensitivity)'].values[0]),
        "false_alarm_rate": float(metrics_df[metrics_df['Label'] == 'Default (0.5)']['False Alarm Rate'].values[0]),
        "auc": float(roc_auc)
    },
    "analysis_date": pd.Timestamp.now().strftime('%Y-%m-%d')
}

# Save findings as JSON
findings_file = ANALYSIS_DIR / 'pattern_analysis_findings.json'
with open(findings_file, 'w') as f:
    json.dump(findings, f, indent=2)

print(f"Analysis findings saved to {findings_file}")

# 10. Summary and Recommendations
print("\n10. Summary and Recommendations:")

print("\nKey Patterns for Fall Detection:")
print("-------------------------------")
print(f"1. Most important features: {', '.join(top_features_list[:5])}")

if len(motion_comparison) > 0:
    distinctive_falls = motion_comparison[motion_comparison['Ratio'] > 1].sort_values('Ratio', ascending=False)
    distinctive_nonfalls = motion_comparison[motion_comparison['Ratio'] < 1].sort_values('Ratio', ascending=True)
    
    if len(distinctive_falls) > 0:
        print(f"2. Distinctive fall features: {', '.join(distinctive_falls['Feature'].head(3))}")
    
    if len(distinctive_nonfalls) > 0:
        print(f"3. Distinctive non-fall features: {', '.join(distinctive_nonfalls['Feature'].head(3))}")

print(f"4. Optimal threshold for balanced performance (F1): {optimal_f1_threshold:.3f}")
print(f"5. Optimal threshold for high sensitivity (F2): {optimal_f2_threshold:.3f}")

print("\nRecommendations:")
print("---------------")
print("1. Threshold Setting:")
if optimal_f2_threshold < 0.4:
    print(f"   Use a lower threshold ({optimal_f2_threshold:.2f}) to prioritize detecting all falls")
    print(f"   This may increase false alarms but minimizes missed falls")
else:
    print(f"   Use the F1-optimal threshold ({optimal_f1_threshold:.2f}) for balanced performance")

print("\n2. Feature Engineering Focus:")
for i, feature in enumerate(top_features_list[:3]):
    print(f"   Optimize detection of {feature}")

print("\n3. Error Reduction Strategy:")
if len(false_negatives) > 0:
    print("   To reduce missed falls:")
    fn_features = false_negatives[top_features_list].mean().sort_values(ascending=False)
    print(f"   - Focus on better detecting {fn_features.index[0]} and {fn_features.index[1]}")

if len(false_positives) > 0:
    print("   To reduce false alarms:")
    fp_features = false_positives[top_features_list].mean().sort_values(ascending=False)
    print(f"   - Improve discrimination of {fp_features.index[0]} and {fp_features.index[1]}")

print("\nPattern analysis complete!")
X
```
