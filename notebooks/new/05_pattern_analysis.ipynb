```python
# 05_pattern_analysis.ipynb - Video Only Version
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import json
import pickle
from sklearn.metrics import confusion_matrix, classification_report, roc_curve, auc, precision_recall_curve
from sklearn.inspection import permutation_importance
from sklearn.model_selection import cross_val_predict
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
import shap

# Set up paths
ROOT_DIR = Path(".")
FEATURES_DIR = ROOT_DIR / "outputs" / "features"
MODELS_DIR = ROOT_DIR / "outputs" / "models"
ANALYSIS_DIR = ROOT_DIR / "outputs" / "analysis" / "video_only"
os.makedirs(ANALYSIS_DIR, exist_ok=True)

print("Fall Detection - Pattern Analysis for Videos Only")

# 1. Load Model and Data
print("\n1. Loading Model and Data...")

# Load the best model
model_file = MODELS_DIR / 'fall_detection_model.pkl'
with open(model_file, 'rb') as f:
    model = pickle.load(f)

# Load model info
info_file = MODELS_DIR / 'model_info.json'
with open(info_file, 'r') as f:
    model_info = json.load(f)

print(f"Loaded model: {model_info['model_name']}")

# Load feature data
def load_feature_data():
    """Load video feature data from CSV with normalized features"""
    normalized_file = FEATURES_DIR / "normalized_features.csv"
    raw_file = FEATURES_DIR / "features_for_modeling.csv"
    
    if normalized_file.exists():
        print(f"Loading normalized features from: {normalized_file}")
        df = pd.read_csv(normalized_file)
    elif raw_file.exists():
        print(f"Loading raw features from: {raw_file}")
        df = pd.read_csv(raw_file)
    else:
        raise FileNotFoundError("No feature data found!")
    
    # Ensure we're only using video data
    if 'data_type' in df.columns:
        df = df[df['data_type'] == 'video']
    
    print(f"Loaded video feature data with {df.shape[0]} samples and {df.shape[1]} features")
    return df

# Load feature data
feature_df = load_feature_data()

# Prepare data for analysis
# Get the exact features the model was trained with
model_features = model_info['features']

# Filter to include only video-specific features if needed
video_features = [feat for feat in model_features if feat in feature_df.columns]

# Use only those features
X = feature_df[video_features]

# Map labels to binary (fall vs non-fall)
if 'label' in feature_df.columns:
    if feature_df['label'].dtype == object:
        y = feature_df['label'].map({'fall': 1, 'workout': 0, 'activity': 0})
    else:
        y = feature_df['label']
else:
    raise ValueError("Label column not found in feature data")

# 2. Identify Distinctive Patterns Between Falls and Workouts
print("\n2. Identifying Distinctive Video Patterns Between Falls and Workouts...")

# Function to plot feature distributions by label
def plot_feature_distributions(df, feature_names, figsize=(15, 10)):
    """Plot distribution of features by class label"""
    num_features = len(feature_names)
    num_rows = (num_features + 2) // 3  # Calculate rows needed (3 columns)
    
    fig, axes = plt.subplots(num_rows, 3, figsize=figsize)
    axes = axes.flatten()
    
    for i, feature in enumerate(feature_names):
        if i < len(axes):
            sns.boxplot(x='label', y=feature, data=df, ax=axes[i])
            axes[i].set_title(f'Distribution of {feature}')
            axes[i].set_xlabel('')
            if df['label'].dtype == 'object':
                axes[i].set_xticklabels(['Non-Fall', 'Fall'])
            else:
                axes[i].set_xticklabels(['Non-Fall', 'Fall'])
    
    # Hide unused subplots
    for i in range(num_features, len(axes)):
        axes[i].set_visible(False)
    
    plt.tight_layout()
    plt.suptitle('Video Feature Distributions: Falls vs Non-Falls', y=1.02)
    return fig

# Get top features based on model info or feature importance
top_features = model_info.get('top_features', [])

# If top features not found in model info, try to extract them
if not top_features and hasattr(model[-1], 'feature_importances_'):
    importances = model[-1].feature_importances_
    indices = np.argsort(importances)[::-1]
    top_features = [X.columns[i] for i in indices[:10]]
elif not top_features:
    # If we can't get feature importances, use some key fall-related video features
    fall_related_features = [
        'max_upper_body_velocity_y', 'max_upper_body_accel_y',
        'height_reduction_pct', 'aspect_ratio_increase_pct',
        'max_height_collapse_velocity', 'max_deceleration',
        'has_impact', 'impact_count', 'velocity_change_ratio',
        'motion_energy_ratio'
    ]
    # Filter to include only features that exist in our dataset
    top_features = [f for f in fall_related_features if f in X.columns][:10]

# Create a DataFrame with features and label for analysis
analysis_df = pd.DataFrame(X[top_features])
analysis_df['label'] = y

# Plot distribution of top features
print("Plotting distributions of top video features...")
dist_fig = plot_feature_distributions(analysis_df, top_features)
dist_fig.savefig(ANALYSIS_DIR / 'video_feature_distributions.png')
plt.show()

# Calculate correlation between features
correlation_matrix = analysis_df.corr()

# Plot correlation heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Video Feature Correlation Matrix')
plt.tight_layout()
plt.savefig(ANALYSIS_DIR / 'video_feature_correlation.png')
plt.show()

# Create a pairplot for top 5 features
if len(top_features) >= 5:
    top_5_features = top_features[:5]
    pair_df = analysis_df[top_5_features + ['label']].copy()
    if pair_df['label'].dtype != 'object':
        pair_df['label'] = pair_df['label'].map({0: 'Non-Fall', 1: 'Fall'})
    
    print("Creating pairplot for top 5 video features...")
    pair_plot = sns.pairplot(pair_df, hue='label', diag_kind='kde', 
                           plot_kws={'alpha': 0.6})
    pair_plot.fig.suptitle('Pairwise Relationships Between Top Video Features', y=1.02)
    pair_plot.savefig(ANALYSIS_DIR / 'video_feature_pairplot.png')
    plt.show()

# 3. Analyze False Positives and False Negatives
print("\n3. Analyzing False Positives and False Negatives in Video Detection...")

# Get model predictions
y_pred = model.predict(X)
y_prob = model.predict_proba(X)[:, 1] if hasattr(model, 'predict_proba') else None

# Create a DataFrame for error analysis
error_df = pd.DataFrame({
    'true_label': y,
    'predicted_label': y_pred,
    'probability': y_prob if y_prob is not None else np.nan
})

# Add features for analysis
for feature in top_features:
    error_df[feature] = X[feature]

# Identify false positives and false negatives
false_positives = error_df[(error_df['true_label'] == 0) & (error_df['predicted_label'] == 1)]
false_negatives = error_df[(error_df['true_label'] == 1) & (error_df['predicted_label'] == 0)]
true_positives = error_df[(error_df['true_label'] == 1) & (error_df['predicted_label'] == 1)]
true_negatives = error_df[(error_df['true_label'] == 0) & (error_df['predicted_label'] == 0)]

print(f"Correctly classified falls (TP): {len(true_positives)}")
print(f"Correctly classified workouts (TN): {len(true_negatives)}")
print(f"Workouts misclassified as falls (FP): {len(false_positives)}")
print(f"Missed falls (FN): {len(false_negatives)}")

# Calculate confusion matrix
cm = confusion_matrix(y, y_pred)

# Plot confusion matrix with percentages
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
           xticklabels=['Non-Fall', 'Fall'],
           yticklabels=['Non-Fall', 'Fall'])
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Video-Only Confusion Matrix')
plt.savefig(ANALYSIS_DIR / 'video_confusion_matrix.png')
plt.show()

# Analyze false positives
if len(false_positives) > 0:
    print("\nAnalyzing False Positives (Workouts classified as falls):")
    print(false_positives[top_features].describe())
    
    # Compare false positives to true positives
    fp_vs_tp = pd.DataFrame({
        'False Positive': false_positives[top_features].mean(),
        'True Positive': true_positives[top_features].mean() if len(true_positives) > 0 else pd.Series([np.nan] * len(top_features), index=top_features)
    })
    
    # Plot comparison
    plt.figure(figsize=(12, 8))
    fp_vs_tp.plot(kind='bar')
    plt.title('Video Feature Comparison: False Positives vs True Positives (Mean Values)')
    plt.ylabel('Normalized Feature Value')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'video_false_positives_analysis.png')
    plt.show()

# Analyze false negatives
if len(false_negatives) > 0:
    print("\nAnalyzing False Negatives (Falls classified as workouts):")
    print(false_negatives[top_features].describe())
    
    # Compare false negatives to true negatives
    fn_vs_tn = pd.DataFrame({
        'False Negative': false_negatives[top_features].mean(),
        'True Negative': true_negatives[top_features].mean() if len(true_negatives) > 0 else pd.Series([np.nan] * len(top_features), index=top_features)
    })
    
    # Plot comparison
    plt.figure(figsize=(12, 8))
    fn_vs_tn.plot(kind='bar')
    plt.title('Video Feature Comparison: False Negatives vs True Negatives (Mean Values)')
    plt.ylabel('Normalized Feature Value')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'video_false_negatives_analysis.png')
    plt.show()

# 4. Feature Importance Assessment
print("\n4. Detailed Video Feature Importance Assessment...")

# Different approaches based on model type
if hasattr(model[-1], 'feature_importances_'):
    # For tree-based models like Random Forest or Gradient Boosting
    feature_importances = model[-1].feature_importances_
    importance_df = pd.DataFrame({
        'Feature': X.columns,
        'Importance': feature_importances
    }).sort_values('Importance', ascending=False)
    
    # Plot feature importances
    plt.figure(figsize=(12, 8))
    sns.barplot(x='Importance', y='Feature', data=importance_df.head(20))
    plt.title('Top 20 Video Features by Importance')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'video_feature_importance.png')
    plt.show()
    
    print("Top 20 most important video features:")
    print(importance_df.head(20).to_string(index=False))
    
    # Save feature importances to CSV
    importance_df.to_csv(ANALYSIS_DIR / 'video_feature_importances.csv', index=False)
    
    # Try to use SHAP values for more detailed feature importance
    try:
        print("\nCalculating SHAP values for video feature interpretation...")
        explainer = shap.TreeExplainer(model[-1])
        shap_values = explainer.shap_values(X)
        
        if isinstance(shap_values, list):
            # For multiclass models
            shap_values = shap_values[1]  # Use values for the fall class
        
        # Summary plot
        plt.figure(figsize=(10, 12))
        shap.summary_plot(shap_values, X, plot_type="bar", show=False)
        plt.title('Video Feature Importance based on SHAP Values')
        plt.tight_layout()
        plt.savefig(ANALYSIS_DIR / 'video_shap_importance.png')
        plt.show()
        
        # Detailed SHAP plot for top features
        plt.figure(figsize=(12, 10))
        shap.summary_plot(shap_values, X, max_display=10)
        plt.tight_layout()
        plt.savefig(ANALYSIS_DIR / 'video_shap_summary.png')
        plt.show()
    except Exception as e:
        print(f"SHAP analysis failed: {e}")
        print("Continuing with alternative feature importance methods...")
    
    # Perform permutation importance analysis
    print("\nCalculating permutation feature importance for video features...")
    perm_importance = permutation_importance(model, X, y, n_repeats=10, random_state=42)
    
    # Create DataFrame and sort
    perm_importance_df = pd.DataFrame({
        'Feature': X.columns,
        'Importance': perm_importance.importances_mean,
        'Std': perm_importance.importances_std
    }).sort_values('Importance', ascending=False)
    
    # Plot permutation importance
    plt.figure(figsize=(12, 8))
    features = perm_importance_df['Feature'].head(20)
    importance = perm_importance_df['Importance'].head(20)
    std = perm_importance_df['Std'].head(20)
    
    # Create horizontal bar plot
    plt.barh(features, importance, xerr=std, capsize=5)
    plt.xlabel('Importance')
    plt.ylabel('Feature')
    plt.title('Video Permutation Feature Importance (Top 20)')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'video_permutation_importance.png')
    plt.show()
    
    print("Top 20 video features by permutation importance:")
    print(perm_importance_df.head(20).to_string(index=False))
    
elif hasattr(model[-1], 'coef_'):
    # For linear models like SVM with linear kernel
    if len(model[-1].coef_.shape) == 1:
        # Binary classification
        coefficients = model[-1].coef_
    else:
        # Get coefficients for the positive class
        coefficients = model[-1].coef_[1] if model[-1].coef_.shape[0] > 1 else model[-1].coef_[0]
    
    # Create DataFrame with absolute values for sorting
    coef_df = pd.DataFrame({
        'Feature': X.columns,
        'Coefficient': coefficients,
        'Abs_Coefficient': np.abs(coefficients)
    }).sort_values('Abs_Coefficient', ascending=False)
    
    # Plot absolute coefficients
    plt.figure(figsize=(12, 8))
    sns.barplot(x='Abs_Coefficient', y='Feature', data=coef_df.head(20))
    plt.title('Top 20 Video Features by Coefficient Magnitude')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'video_coefficient_magnitude.png')
    plt.show()
    
    # Plot coefficients with direction
    plt.figure(figsize=(12, 8))
    sns.barplot(x='Coefficient', y='Feature', data=coef_df.head(20))
    plt.title('Top 20 Video Features by Coefficient (with Direction)')
    plt.axvline(x=0, color='gray', linestyle='--')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'video_coefficient_direction.png')
    plt.show()
    
    print("Top 20 most influential video features:")
    print(coef_df[['Feature', 'Coefficient']].head(20).to_string(index=False))
    
    # Save coefficients to CSV
    coef_df.to_csv(ANALYSIS_DIR / 'video_feature_coefficients.csv', index=False)
else:
    print("Feature importance assessment not available for this model type.")
    print("Using permutation importance instead...")
    
    # Permutation importance works for any model
    perm_importance = permutation_importance(model, X, y, n_repeats=10, random_state=42)
    
    # Create DataFrame and sort
    perm_importance_df = pd.DataFrame({
        'Feature': X.columns,
        'Importance': perm_importance.importances_mean,
        'Std': perm_importance.importances_std
    }).sort_values('Importance', ascending=False)
    
    # Plot permutation importance
    plt.figure(figsize=(12, 8))
    features = perm_importance_df['Feature'].head(20)
    importance = perm_importance_df['Importance'].head(20)
    std = perm_importance_df['Std'].head(20)
    
    # Create horizontal bar plot
    plt.barh(features, importance, xerr=std, capsize=5)
    plt.xlabel('Importance')
    plt.ylabel('Feature')
    plt.title('Video Permutation Feature Importance (Top 20)')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'video_permutation_importance.png')
    plt.show()
    
    print("Top 20 video features by permutation importance:")
    print(perm_importance_df.head(20).to_string(index=False))

# 5. Threshold Tuning for Real-World Application
print("\n5. Threshold Tuning for Real-World Video-Based Fall Detection...")

# Check if we can get probabilities
if hasattr(model, 'predict_proba'):
    # Get predicted probabilities
    y_prob = model.predict_proba(X)[:, 1]
    
    # Calculate precision and recall at different thresholds
    precision, recall, thresholds = precision_recall_curve(y, y_prob)
    
    # Calculate F1 scores at different thresholds
    f1_scores = 2 * (precision[:-1] * recall[:-1]) / (precision[:-1] + recall[:-1] + 1e-10)
    
    # Calculate F2 scores (weighing recall higher)
    beta = 2
    f2_scores = (1 + beta**2) * (precision[:-1] * recall[:-1]) / ((beta**2 * precision[:-1]) + recall[:-1] + 1e-10)
    
    # Create DataFrame for threshold analysis
    threshold_df = pd.DataFrame({
        'Threshold': thresholds,
        'Precision': precision[:-1],
        'Recall': recall[:-1],
        'F1': f1_scores,
        'F2': f2_scores
    })
    
    # Plot precision-recall curve
    plt.figure(figsize=(10, 6))
    plt.plot(recall, precision, marker='.', label='Precision-Recall curve')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title('Video-Based Fall Detection Precision-Recall Curve')
    plt.grid(True)
    plt.savefig(ANALYSIS_DIR / 'video_precision_recall_curve.png')
    plt.show()
    
    # Plot F1 and F2 scores vs threshold
    plt.figure(figsize=(10, 6))
    plt.plot(thresholds, f1_scores, label='F1 Score')
    plt.plot(thresholds, f2_scores, label='F2 Score (weighs recall higher)')
    plt.xlabel('Threshold')
    plt.ylabel('Score')
    plt.title('Video-Based F1 and F2 Scores vs. Threshold')
    plt.legend()
    plt.grid(True)
    plt.savefig(ANALYSIS_DIR / 'video_f_scores_vs_threshold.png')
    plt.show()
    
    # Find optimal thresholds
    optimal_f1_idx = np.argmax(f1_scores)
    optimal_f1_threshold = thresholds[optimal_f1_idx]
    
    optimal_f2_idx = np.argmax(f2_scores)
    optimal_f2_threshold = thresholds[optimal_f2_idx]
    
    print(f"Default classification threshold: 0.5")
    print(f"Optimal threshold for balanced precision/recall (F1): {optimal_f1_threshold:.4f}")
    print(f"Optimal threshold for higher recall (F2): {optimal_f2_threshold:.4f}")
    
    # Calculate confusion matrices for different thresholds
    cms = {}
    threshold_labels = {
        0.5: 'Default (0.5)',
        optimal_f1_threshold: 'F1 Optimal',
        optimal_f2_threshold: 'F2 Optimal (Recall Focus)'
    }
    
    plt.figure(figsize=(15, 5))
    for i, (threshold, label) in enumerate(threshold_labels.items()):
        y_pred_custom = (y_prob >= threshold).astype(int)
        cm = confusion_matrix(y, y_pred_custom)
        cms[label] = cm
        
        # Plot confusion matrix
        plt.subplot(1, 3, i+1)
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=['Non-Fall', 'Fall'],
                   yticklabels=['Non-Fall', 'Fall'])
        plt.xlabel('Predicted')
        plt.ylabel('True')
        plt.title(f'Threshold: {threshold:.3f} ({label})')
    
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'video_threshold_confusion_matrices.png')
    plt.show()
    
    # Calculate metrics for different thresholds
    metrics = []
    for threshold, label in threshold_labels.items():
        y_pred_custom = (y_prob >= threshold).astype(int)
        cm = confusion_matrix(y, y_pred_custom)
        
        # Calculate metrics
        tn, fp, fn, tp = cm.ravel()
        precision_val = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall_val = tp / (tp + fn) if (tp + fn) > 0 else 0
        specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
        f1 = 2 * (precision_val * recall_val) / (precision_val + recall_val) if (precision_val + recall_val) > 0 else 0
        accuracy = (tp + tn) / (tp + fp + tn + fn)
        
        # For real-world fall detection, add more metrics
        false_alarm_rate = fp / (fp + tn) if (fp + tn) > 0 else 0
        miss_rate = fn / (fn + tp) if (fn + tp) > 0 else 0
        
        metrics.append({
            'Threshold': threshold,
            'Label': label,
            'Accuracy': accuracy,
            'Precision': precision_val,
            'Recall (Sensitivity)': recall_val,
            'Specificity': specificity,
            'F1 Score': f1,
            'False Alarm Rate': false_alarm_rate,
            'Miss Rate': miss_rate,
            'TP': tp,
            'FP': fp,
            'TN': tn,
            'FN': fn
        })
    
    # Create metrics DataFrame
    metrics_df = pd.DataFrame(metrics)
    print("\nPerformance metrics at different thresholds for video-based detection:")
    print(metrics_df[['Label', 'Threshold', 'Accuracy', 'Precision', 'Recall (Sensitivity)', 'False Alarm Rate', 'Miss Rate']])
    
    # Save metrics to CSV
    metrics_df.to_csv(ANALYSIS_DIR / 'video_threshold_metrics.csv', index=False)
    
    # Create a more detailed analysis of the optimal thresholds
    plt.figure(figsize=(12, 6))
    
    # Plot precision, recall vs threshold
    plt.plot(thresholds, precision[:-1], label='Precision', marker='', linestyle='-')
    plt.plot(thresholds, recall[:-1], label='Recall', marker='', linestyle='-')
    
    # Highlight optimal thresholds
    plt.axvline(x=0.5, color='gray', linestyle='--', label='Default (0.5)')
    plt.axvline(x=optimal_f1_threshold, color='green', linestyle='--', label=f'F1 Optimal ({optimal_f1_threshold:.3f})')
    plt.axvline(x=optimal_f2_threshold, color='red', linestyle='--', label=f'F2 Optimal ({optimal_f2_threshold:.3f})')
    
    plt.xlabel('Threshold')
    plt.ylabel('Score')
    plt.title('Video-Based Precision & Recall vs. Threshold')
    plt.legend()
    plt.grid(True)
    plt.savefig(ANALYSIS_DIR / 'video_threshold_analysis.png')
    plt.show()
    
    # ROC curve and AUC
    fpr, tpr, _ = roc_curve(y, y_prob)
    roc_auc = auc(fpr, tpr)
    
    plt.figure(figsize=(10, 6))
    plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC curve (AUC = {roc_auc:.3f})')
    plt.plot([0, 1], [0, 1], color='gray', lw=2, linestyle='--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Video-Based ROC Curve')
    plt.legend(loc="lower right")
    plt.grid(True)
    plt.savefig(ANALYSIS_DIR / 'video_roc_curve.png')
    plt.show()
    
    # Recommendation section
    print("\nThreshold Recommendations for Real-World Video-Based Fall Detection:")
    
    # Based on metrics, provide recommendations
    if metrics_df.loc[metrics_df['Label'] == 'F2 Optimal (Recall Focus)', 'Miss Rate'].values[0] < 0.1:
        print("RECOMMENDED THRESHOLD: F2 Optimal")
        print(f"Set threshold to {optimal_f2_threshold:.4f} for high fall detection rate (priority to minimize missed falls)")
        recommended_threshold = optimal_f2_threshold
    else:
        max_recall_idx = np.argmax(recall[:-1])
        high_recall_threshold = thresholds[max_recall_idx]
        print("RECOMMENDED THRESHOLD: Maximum Recall")
        print(f"Set threshold to {high_recall_threshold:.4f} for maximum fall detection rate")
        recommended_threshold = high_recall_threshold
    
    print("\nVideo-based fall detection priority recommendations:")
    print(f"1. High Sensitivity Setting (minimize missed falls): {recommended_threshold:.4f}")
    print(f"2. Balanced Setting (F1 optimal): {optimal_f1_threshold:.4f}")
    print(f"3. High Specificity Setting (minimize false alarms): 0.7")
    
    # Save the recommended thresholds to a JSON file
    threshold_recommendations = {
        'default': 0.5,
        'f1_optimal': float(optimal_f1_threshold),
        'f2_optimal': float(optimal_f2_threshold),
        'recommended': float(recommended_threshold),
        'high_specificity': 0.7
    }
    
    with open(ANALYSIS_DIR / 'video_threshold_recommendations.json', 'w') as f:
        json.dump(threshold_recommendations, f, indent=2)
    
    print(f"\nVideo threshold recommendations saved to {ANALYSIS_DIR / 'video_threshold_recommendations.json'}")
else:
    print("This model doesn't support predict_proba, so threshold tuning is not possible")

# 6. Video-Specific Fall vs. Workout Motion Pattern Analysis
print("\n6. Video-Specific Fall vs. Workout Motion Pattern Analysis...")

# Select video-specific motion features
motion_features = [col for col in X.columns if any(term in col for term in 
                                                  ['velocity', 'accel', 'height', 'collapse',
                                                   'movement', 'aspect_ratio', 'bounding_box'])]

if motion_features:
    # Filter to meaningful features
    motion_features = motion_features[:8]  # Limit to 8 for clarity
    
    # Create a detailed visualization of motion patterns
    plt.figure(figsize=(15, 10))
    for i, feature in enumerate(motion_features):
        if i < 8:  # Ensure we only plot up to 8 features
            plt.subplot(2, 4, i+1)
            if feature in X.columns:
                sns.boxplot(x='label', y=feature, data=pd.DataFrame({
                    'label': y.map({0: 'Non-Fall', 1: 'Fall'}),
                    feature: X[feature]
                }))
                plt.title(f'{feature}')
                plt.xlabel('')
    
    plt.tight_layout()
    plt.suptitle('Video Motion Features: Falls vs Non-Falls', y=1.02)
    plt.savefig(ANALYSIS_DIR / 'video_motion_analysis.png')
    plt.show()
    
    # Create a radar chart to visualize fall vs. workout motion patterns
    fall_means = pd.DataFrame(X[X.index.isin(true_positives.index)][motion_features].mean())
    nonfall_means = pd.DataFrame(X[X.index.isin(true_negatives.index)][motion_features].mean())
    
    # Normalize values for radar chart
    combined_df = pd.DataFrame({
        'Fall': fall_means[0],
        'Non-Fall': nonfall_means[0]
    })
    
    # Plot side-by-side comparison
    plt.figure(figsize=(12, 8))
    combined_df.plot(kind='bar')
    plt.title('Fall vs. Non-Fall Video Motion Feature Comparison')
    plt.ylabel('Feature Value')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(ANALYSIS_DIR / 'video_fall_pattern_comparison.png')
    plt.show()
    
    # Calculate and display the ratio between falls and non-falls
    combined_df['Ratio'] = combined_df['Fall'] / combined_df['Non-Fall']
    print("\nRatio of fall to non-fall feature values:")
    print(combined_df['Ratio'].sort_values(ascending=False))
    
    # Identify the most distinctive fall features
    print("\nMost distinctive fall features (highest fall/non-fall ratio):")
    for i, (feature, ratio) in enumerate(combined_df['Ratio'].sort_values(ascending=False).items()):
        if i < 5:  # Show top 5
            print(f"{feature}: {ratio:.2f}x higher in falls")
    
    # Identify features where non-falls have higher values
    print("\nFeatures where non-falls have higher values:")
    for feature, ratio in combined_df['Ratio'][combined_df['Ratio'] < 1].sort_values().items():
        print(f"{feature}: {1/ratio:.2f}x higher in non-falls")
```
